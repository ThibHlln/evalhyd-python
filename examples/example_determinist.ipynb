```python
import numpy
import evalhyd
```

## Calculate Nash-Sutcliffe efficiency on 1D arrays

```python
# streamflow observations
obs = numpy.array(
    [4.7, 4.3, 5.5, 2.7]
)
# streamflow predictions
prd = numpy.array(
    [5.3, 4.2, 5.7, 2.3]
)

print(obs.shape, prd.shape)
```

```python
nse, = evalhyd.evald(obs, prd, ['NSE'])
print(nse)
```

## Calculate Nash-Sutcliffe efficiency on 2D arrays

```python
# streamflow observations
obs = numpy.array(
    [[4.7, 4.3, 5.5, 2.7]]
)
# streamflow predictions
prd = numpy.array(
    [[5.3, 4.2, 5.7, 2.3],
     [4.3, 4.2, 4.7, 4.3],
     [5.3, 5.2, 5.7, 2.3]]
)

print(obs.shape, prd.shape)
```

```python
nse, = evalhyd.evald(obs, prd, ['NSE'])
print(nse)
```
