```python
import numpy
import evalhyd
```

```python
sim = numpy.array(
    [5.3, 4.2, 5.7, 2.3]
)
obs = numpy.array(
    [4.7, 4.3, 5.5, 2.7]
)
print(evalhyd.nse(sim, obs))
```

```python
sim = numpy.array(
    [[5.3, 5.3, 5.3],
     [4.2, 4.2, 4.2],
     [5.7, 5.7, 5.7],
     [2.3, 2.3, 2.3]]
)
obs = numpy.array(
    [[4.7],
     [4.3],
     [5.5],
     [2.7]]
)
print(sim.shape, obs.shape)
print(evalhyd.nse(sim, obs, axis=0))
```

```python
sim = numpy.array(
     [[5.3, 4.2, 5.7, 2.3],
      [5.3, 4.2, 5.7, 2.3],
      [5.3, 4.2, 5.7, 2.3]]
)
obs = numpy.array(
    [[4.7, 4.3, 5.5, 2.7]]
)
print(sim.shape, obs.shape)
print(evalhyd.nse(sim, obs, axis=1))
```
